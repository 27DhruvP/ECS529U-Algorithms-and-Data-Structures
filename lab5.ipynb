## Question 1 [2]

Add to `ArrayList` a function

    def appendAll(self, A)
    
that appends all elements of the array `A` in the array list (the one represented by `self`). 

For example, if `ls` is `[2,3,4,5]` then `ls.appendAll([42,24])` changes `ls` to `[2,3,4,5,42,24]`.

Then, add to `ArrayList` a function

    def toArray(self)
    
that returns a new array containing the same elements as the array list represented by 
`self`. Note that the length of the array should be the same as that of the array list, and not 
the same as that of its internal array. 

```python
class ArrayList:
    def __init__(self):
        self.inArray = [0 for i in range(10)]
        self.count = 0
        
    def get(self, i):
        return self.inArray[i]

    def set(self, i, e):
        self.inArray[i] = e

    def length(self):
        return self.count

    def append(self, e):
        self.inArray[self.count] = e
        self.count += 1
        if len(self.inArray) == self.count:
            self._resizeUp()     # resize array if reached capacity

    def insert(self, i, e):
        for j in range(self.count,i,-1):
            self.inArray[j] = self.inArray[j-1]
        self.inArray[i] = e
        self.count += 1
        if len(self.inArray) == self.count:
            self._resizeUp()     # resize array if reached capacity

    def remove(self, i):
        self.count -= 1
        val = self.inArray[i]
        for j in range(i,self.count):
            self.inArray[j] = self.inArray[j+1]
        return val
    
    def __str__(self):
        return str(self.inArray[:self.count])

    def _resizeUp(self):
        newArray = [0 for i in range(2*len(self.inArray))]
        for j in range(len(self.inArray)):
            newArray[j] = self.inArray[j]
        self.inArray = newArray
        
    def appendAll(self, A):
        for i in range(len(A)):
            self.append(A[i])

    def toArray(self):
        return self.inArray[:self.count]

```

## Priority Queues

For the next question, we look at another data structure, namely priority queues. A priority 
queue is a queue in which each element has a priority, and where dequeueing always 
returns the item with the greatest priority in the queue.

We start by defining a class of priority queue elements (PQ-elements for short):

    class PQElement:
        def __init__(self, v, p):
            self.val = v
            self.priority = p
            
So, a PQ-element is a pair consisting of a value (which can be anything, e.g. an integer, a 
string, an array, etc.) and a priority (which is an integer). 

Below we also implemented the `__str__` function to be able to print PQ-elements.

```python
class PQElement:
    def __init__(self, v, p):
        self.val = v
        self.priority = p
    
    def __str__(self):
        return str((self.val,self.priority))
```

```python
# ===== Test cell. Code was produced with Claude =====
# ===== This cell should be AFTER your solutions (Questions 1, 2, 6, 7 and 8) =====

import random, sys
from itertools import product

# ---------- helpers ----------

def _elems(ls):
    # the elements of an array list, read directly from its internal array
    return list(ls.inArray[:ls.count])

def _make(A):
    # build an array list using only append (from the provided class)
    ls = ArrayList()
    for x in A:
        ls.append(x)
    return ls

def _make_garbage(A, garbage):
    # an array list with elements A, whose internal array still holds old
    # values (garbage) after position len(A), as happens after removing elements
    ls = _make(A + garbage)
    ls.count = len(A)
    return ls

def _raises(f):
    try:
        f()
    except Exception:
        return True
    return False

def _short(x, n=60):
    s = str(x)
    return s if len(s) <= n else s[:n - 3] + "..."

# Each test is a pair (description, function). The function returns True if the
# test passes, or a string explaining what went wrong.

# ---------- Question 1: appendAll(self, A) and toArray(self) ----------

def _q1_appendAll(start, A):
    def run():
        ls = _make(start)
        A2 = A[:]
        ls.appendAll(A2)
        if A2 != A:
            return "the array A was changed"
        expected = start + A
        if _elems(ls) != expected or ls.length() != len(expected):
            return f"the list became {_short(_elems(ls))}, expected {_short(expected)}"
        if A2:
            A2[0] = -12345              # changing A afterwards must not change the list
            if _elems(ls) != expected:
                return "the list shares its internal array with A (changing A changed the list)"
        ls.append(99)                   # the list must still work normally afterwards
        if _elems(ls) != expected + [99]:
            return "the list does not work correctly after appendAll (append gave a wrong result)"
        return True
    return (f"{_short(start, 30)}.appendAll({_short(A, 30)})", run)

def _q1_toArray(A, garbage=[]):
    def run():
        ls = _make_garbage(A, garbage)
        out = ls.toArray()
        if not isinstance(out, list):
            return f"returned a {type(out).__name__}, expected an array (Python list)"
        if out != A:
            return f"returned {_short(out)}, expected {_short(A)}"
        if _elems(ls) != A:
            return "the array list was changed"
        if out:
            out[0] = -12345             # the result must be a NEW array
            if _elems(ls) != A:
                return "the returned array is not a new array (changing it changed the list)"
        return True
    return (f"toArray of {_short(A, 40)}" + (" (with old values left in the internal array)" if garbage else ""), run)

TESTS_Q1 = [_q1_appendAll(s, A) for s, A in [
    ([2, 3, 4, 5], [42, 24]),           # example from the lab sheet
    ([], []),                           # empty list, empty array
    ([], [1]),
    ([1, 2], []),                       # nothing to append
    ([], list(range(25))),              # needs the internal array to grow
    ([0] * 9, [1, 2, 3]),               # crosses the internal capacity of 10
    (list(range(15)), list(range(100, 140))),
    ([-1, 0], [5, 5, 5]),
]]
TESTS_Q1 += [_q1_toArray(A) for A in [[], [7], [1, 2, 3, 4, 5], list(range(9)), list(range(10)),
                                       list(range(11)), list(range(25)), [-3, 0, -3]]]
TESTS_Q1 += [_q1_toArray([1, 2, 3], [9, 9]), _q1_toArray([], [4, 4, 4])]

# ---------- Question 2: get, set, remove and insert check their index ----------
# Valid indices: 0..length-1 for get, set and remove; 0..length for insert.
# Out-of-bounds indices (including negative ones) must raise an exception and
# leave the list unchanged.

_CALLS = {
    "get":    lambda ls, i: ls.get(i),
    "set":    lambda ls, i: ls.set(i, 777),
    "remove": lambda ls, i: ls.remove(i),
    "insert": lambda ls, i: ls.insert(i, 777),
}

def _q2_bad(op, A, i, garbage=[]):
    def run():
        ls = _make_garbage(A, garbage)
        before = _elems(ls)
        if not _raises(lambda: _CALLS[op](ls, i)):
            return f"no exception was raised for index {i} on a list of length {len(A)}"
        if _elems(ls) != before:
            return f"the list was changed (to {_short(_elems(ls))}) although an exception was raised"
        return True
    return (f"{op}({i}) on {_short(A, 40)} raises an exception", run)

def _q2_good(op, A, i, garbage=[]):
    def run():
        ls = _make_garbage(A, garbage)
        out = _CALLS[op](ls, i)
        if op == "get":
            exp_list, exp_out = A, A[i]
        elif op == "set":
            exp_list, exp_out = A[:i] + [777] + A[i + 1:], None
        elif op == "remove":
            exp_list, exp_out = A[:i] + A[i + 1:], A[i]
        else:
            exp_list, exp_out = A[:i] + [777] + A[i:], None
        if op in ("get", "remove") and out != exp_out:
            return f"returned {out!r}, expected {exp_out!r}"
        if _elems(ls) != exp_list:
            return f"the list became {_short(_elems(ls))}, expected {_short(exp_list)}"
        return True
    return (f"{op}({i}) on {_short(A, 40)} works", run)

_L5 = [10, 20, 30, 40, 50]      # internal array of length 10, so indices 5..9 hold garbage
_L15 = list(range(15))          # internal array of length 20

TESTS_Q2 = []
for op in ("get", "set", "remove"):
    TESTS_Q2 += [_q2_good(op, _L5, i) for i in (0, 2, 4)]
    TESTS_Q2 += [_q2_bad(op, _L5, i) for i in (-1, -5, 5, 6, 9)]
    TESTS_Q2 += [_q2_bad(op, [], i) for i in (0, -1)]
    TESTS_Q2 += [_q2_good(op, _L15, 14), _q2_bad(op, _L15, 15), _q2_bad(op, _L15, 19)]
    TESTS_Q2 += [_q2_good(op, [1, 2, 3], 2, [9, 9]), _q2_bad(op, [1, 2, 3], 3, [9, 9]),
                 _q2_bad(op, [1, 2, 3], 4, [9, 9])]
TESTS_Q2 += [_q2_good("insert", _L5, i) for i in (0, 2, 5)]           # 5 = at the end: allowed
TESTS_Q2 += [_q2_bad("insert", _L5, i) for i in (-1, -5, 6, 9)]
TESTS_Q2 += [_q2_good("insert", [], 0), _q2_bad("insert", [], 1), _q2_bad("insert", [], -1)]
TESTS_Q2 += [_q2_good("insert", list(range(9)), 9), _q2_good("insert", list(range(9)), 0)]  # needs resizing
TESTS_Q2 += [_q2_good("insert", _L15, 15), _q2_bad("insert", _L15, 16), _q2_bad("insert", _L15, 19)]
TESTS_Q2 += [_q2_good("insert", [1, 2, 3], 3, [9, 9]), _q2_bad("insert", [1, 2, 3], 4, [9, 9])]

# ---------- Question 6: groupScripts(A) ----------
# The tests use their own script class with the same fields (sid, mark) as Script.

class _Script:
    def __init__(self, s, m):
        self.sid = s
        self.mark = m
    def __str__(self):
        return "Script" + str((self.sid, self.mark))

def _fmt_scripts(S):
    return "[" + ", ".join(str(s) for s in S) + "]"

def _q6_check(A, G):
    try:
        n = len(G)
    except Exception:
        return f"returned {G!r}, expected an array of length 101"
    if n != 101:
        return f"returned an array of length {n}, expected 101"
    if G is A:
        return "returned the input array instead of a new one"
    if len(set(id(g) for g in G)) != 101:
        return "the same array list appears in more than one position of G"
    for x in range(101):
        if not isinstance(G[x], ArrayList):
            return f"G[{x}] is a {type(G[x]).__name__}, expected an ArrayList"
        got = _elems(G[x])
        exp = [s for s in A if s.mark == x]
        if sorted(map(id, got)) != sorted(map(id, exp)):
            return f"G[{x}] is {_short(_fmt_scripts(got))}, expected {_short(_fmt_scripts(exp))} (in any order)"
    return True

def _q6_case(pairs, label):
    def run():
        A = [_Script(s, m) for s, m in pairs]
        A2 = A[:]
        G = groupScripts(A2)
        if [id(s) for s in A2] != [id(s) for s in A] or [s.mark for s in A] != [m for _, m in pairs]:
            return "the array A was changed"
        return _q6_check(A, G)
    return (label, run)

def _q6_two_calls():
    A = [_Script(1, 50), _Script(2, 60)]
    B = [_Script(3, 50), _Script(4, 70)]
    G1 = groupScripts(A)
    G2 = groupScripts(B)
    r = _q6_check(B, G2)
    if r is not True:
        return "second call: " + r
    return _q6_check(A, G1)

_rng6 = random.Random()
_Q6_EXAMPLE = [(101, 52), (95, 42), (102, 54), (100, 42), (113, 54), (99, 42)]
TESTS_Q6 = [
    _q6_case(_Q6_EXAMPLE, "example from the lab sheet"),
    _q6_case([], "empty array"),
    _q6_case([(1, 0)], "one script with mark 0"),
    _q6_case([(1, 100)], "one script with mark 100"),
    _q6_case([(1, 0), (2, 100), (3, 0)], "marks 0 and 100"),
    _q6_case([(i, 70) for i in range(25)], "25 scripts with the same mark"),
    _q6_case([(i, i) for i in range(101)], "every mark exactly once"),
    _q6_case(_rng6.sample([(i, i % 101) for i in range(202)], 202), "every mark twice, shuffled"),
    _q6_case([(i, _rng6.randint(0, 100)) for i in range(300)], "300 scripts with random marks"),
    _q6_case([(i, _rng6.choice([0, 40, 100])) for i in range(60)], "60 scripts, marks 0, 40 or 100"),
    ("two calls give separate results", _q6_two_calls),
]

# ---------- Question 7: PQueue ----------
# The order of elements with EQUAL priority is not specified, so any order of
# those is accepted. str must list the elements by decreasing priority, written
# as in the lab sheet, e.g. [('Q', 8),('D', 7)] (spaces are ignored).

def _nospace(s):
    return "".join(str(s).split())

def _str_ok(s, pairs):
    # is s a valid printout of a priority queue containing pairs (value, priority)?
    s = _nospace(s)
    if not (s.startswith("[") and s.endswith("]")):
        return False
    body, pos = s[1:-1], 0
    remaining = [(p, _nospace(PQElement(v, p))) for v, p in pairs]
    while remaining:
        top = max(p for p, _ in remaining)
        match = next((t for p, t in remaining if p == top and body.startswith(t, pos)), None)
        if match is None:
            return False
        remaining.remove((top, match))
        pos += len(match)
        if remaining:
            if body[pos:pos + 1] != ",":
                return False
            pos += 1
    return pos == len(body)

def _q7_deq_check(q, remaining):
    e = q.deq()
    if not (hasattr(e, "val") and hasattr(e, "priority")):
        return f"deq returned {e!r}, expected a PQElement"
    top = max(p for _, p in remaining)
    if e.priority != top:
        return f"deq returned {e}, but an element with priority {top} was in the queue"
    if (e.val, e.priority) not in remaining:
        return f"deq returned {e}, which is not in the queue"
    remaining.remove((e.val, e.priority))
    if q.size() != len(remaining):
        return f"after deq, size() is {q.size()}, expected {len(remaining)}"
    return True

def _q7_seq(pairs, label):
    def run():
        q = PQueue()
        for i, (v, p) in enumerate(pairs):
            q.enq(PQElement(v, p))
            if q.size() != i + 1:
                return f"after {i + 1} enq's, size() is {q.size()}"
        s = str(q)
        if not _str_ok(s, pairs):
            return f"str gave {_short(s, 80)}, expected the elements by decreasing priority"
        if q.size() != len(pairs):
            return "printing the queue changed its size"
        remaining = list(pairs)
        while remaining:
            r = _q7_deq_check(q, remaining)
            if r is not True:
                return r
        if _nospace(q) != "[]":
            return f"after dequeueing everything, str gave {_short(q)}, expected []"
        return True
    return (label, run)

def _q7_example():
    q = PQueue()
    A = [("D", 7), ("S", 5), ("A", 0), ("G", 4), ("Q", 8), ("P", 3), ("A", -4), ("S", 1), ("S", -1), ("T", 2), ("G", -2)]
    for v, p in A:
        q.enq(PQElement(v, p))
    exp = "[('Q', 8),('D', 7),('S', 5),('G', 4),('P', 3),('T', 2),('S', 1),('A', 0),('S', -1),('G', -2),('A', -4)]"
    if _nospace(q) != _nospace(exp):
        return f"str gave {_short(q, 80)}, expected {_short(exp, 80)}"
    e = q.deq()
    if not hasattr(e, "val") or (e.val, e.priority) != ("Q", 8):
        return f"deq returned {e}, expected ('Q', 8)"
    exp = "[('D', 7),('S', 5),('G', 4),('P', 3),('T', 2),('S', 1),('A', 0),('S', -1),('G', -2),('A', -4)]"
    if _nospace(q) != _nospace(exp):
        return f"after deq, str gave {_short(q, 80)}, expected {_short(exp, 80)}"
    return True

def _q7_empty():
    q = PQueue()
    if q.size() != 0:
        return f"size() of a new queue is {q.size()}, expected 0"
    if _nospace(q) != "[]":
        return f"str of a new queue is {q}, expected []"
    return True

def _q7_independent():
    q1, q2 = PQueue(), PQueue()
    q1.enq(PQElement("A", 1))
    if q2.size() != 0 or _nospace(q2) != "[]":
        return "enqueueing in one queue changed another queue"
    return True

def _q7_random(seed):
    def run():
        rng = random.Random(seed)
        q, remaining = PQueue(), []
        for step in range(300):
            if remaining and rng.random() < 0.4:
                r = _q7_deq_check(q, remaining)
                if r is not True:
                    return f"step {step}: {r}"
            else:
                v, p = rng.choice("ABCDEFG"), rng.randint(-5, 5)
                q.enq(PQElement(v, p))
                remaining.append((v, p))
                if q.size() != len(remaining):
                    return f"step {step}: after enq, size() is {q.size()}, expected {len(remaining)}"
            if step % 25 == 0 and not _str_ok(str(q), remaining):
                return f"step {step}: str gave {_short(q, 80)}"
        return True
    return (f"300 random enq/deq operations (with equal priorities), run {seed}", run)

def _lines_in_one_deq(n):
    # build a queue of n elements, then count how many lines of Python code
    # are run by one deq (median of 15 deq's)
    rng = random.Random(n)
    prios = list(range(n))
    rng.shuffle(prios)
    q = PQueue()
    for p in prios:
        q.enq(PQElement("x", p))
    counts = []
    for _ in range(15):
        c = [0]
        def tracer(frame, event, arg):
            if event == "line":
                c[0] += 1
            return tracer
        old = sys.gettrace()
        sys.settrace(tracer)
        try:
            q.deq()
        finally:
            sys.settrace(old)
        counts.append(c[0])
    return sorted(counts)[len(counts) // 2]

def _q7_constant_deq():
    small, big = _lines_in_one_deq(20), _lines_in_one_deq(600)
    if big > small + 25:
        return (f"deq does not take constant time: one deq ran {small} lines of code "
                f"with 20 elements in the queue, but {big} with 600")
    return True

_rng7 = random.Random()
TESTS_Q7 = [
    ("example from the lab sheet", _q7_example),
    ("new queue is empty", _q7_empty),
    ("two queues are independent", _q7_independent),
    _q7_seq([("A", 5)], "one element"),
    _q7_seq([(chr(65 + i), i) for i in range(15)], "15 elements enqueued in increasing priority"),
    _q7_seq([(chr(65 + i), 15 - i) for i in range(15)], "15 elements enqueued in decreasing priority"),
    _q7_seq([(f"v{p}", p) for p in _rng7.sample(range(-50, 50), 30)], "30 elements, random distinct priorities"),
    _q7_seq([("A", -3), ("B", -1), ("C", -7)], "negative priorities"),
    _q7_seq([("A", 1), ("B", 1), ("C", 1)], "all priorities equal"),
    _q7_seq([("A", 1), ("B", 2), ("C", 1), ("D", 2), ("E", 0)], "some priorities equal"),
    _q7_seq([([1, 2], 3), ("word", 9), (42, 6)], "values that are not strings"),
    _q7_random(1), _q7_random(2), _q7_random(3),
    ("deq takes constant time", _q7_constant_deq),
]

# ---------- Question 8: __lt__ (lexicographic order) ----------
# Python already compares lists lexicographically in exactly this way,
# so the expected answer for array lists ls1, ls2 is A1 < A2 for the arrays A1, A2.

def _q8_case(A, B, gA=[], gB=[]):
    def run():
        ls1, ls2 = _make_garbage(A, gA), _make_garbage(B, gB)
        out = ls1 < ls2
        if _elems(ls1) != A or _elems(ls2) != B:
            return "one of the array lists was changed"
        if not isinstance(out, bool) or out != (A < B):
            return f"returned {out!r}, expected {A < B}"
        return True
    extra = " (with old values left in the internal arrays)" if gA or gB else ""
    return (f"{_short(A, 30)} < {_short(B, 30)}{extra}", run)

def _q8_same_object():
    ls = _make([1, 2, 3])
    out = ls < ls
    return True if out is False else f"ls < ls returned {out!r}, expected False"

TESTS_Q8 = [_q8_case(A, B) for A, B in [
    ([2, 2, 3, 4], [2, 2, 5, 1]), ([2, 2, 5, 1], [2, 2, 3, 4]),    # examples from the lab sheet
    ([1, 5, 5, 5], [2, 1, 1, 1]), ([2, 1, 1, 1], [1, 5, 5, 5]),
    ([1, 2, 3], [1, 2, 3, 1]),    ([1, 2, 3, 1], [1, 2, 3]),
    ([], []), ([], [0]), ([0], []), ([-1], []),                     # empty lists
    ([1, 2, 3], [1, 2, 3]),                                        # equal: not smaller
    ([-5, 3], [-4, -9]), (["b", "a"], ["b", "c"]), (["ab"], ["b"]),
]]
_long = list(range(20))
TESTS_Q8 += [_q8_case(_long, _long[:12] + [100] + _long[13:15]), _q8_case(_long[:12] + [100], _long),
             _q8_case(_long, _long[:15]), _q8_case(_long[:15], _long)]
# old values left in the internal array beyond the length must be ignored
TESTS_Q8 += [_q8_case([1, 2, 3], [1, 2, 3, 5], [9, 9]), _q8_case([1, 2, 3, 5], [1, 2, 3], [], [9, 9]),
             _q8_case([1, 2, 3], [1, 2, 3], [9, 9], [0, 0]), _q8_case([1, 2], [1, 2, 5], [1, 2, 5]),
             _q8_case([1, 2, 5], [1, 2], [], [1, 2, 5])]
# every pair of lists of length 0..3 over the values 0, 1, 2
_SMALL = [list(t) for n in range(4) for t in product(range(3), repeat=n)]
TESTS_Q8 += [_q8_case(A, B) for A in _SMALL for B in _SMALL]
TESTS_Q8 += [("ls < ls is False", _q8_same_object)]

# ---------- runner ----------
MAX_SHOWN = 5    # show at most this many failures per question

def run_lab5_tests():
    suites = {1: TESTS_Q1, 2: TESTS_Q2, 6: TESTS_Q6, 7: TESTS_Q7, 8: TESTS_Q8}
    passed = []
    for q, tests in suites.items():
        failures = []
        for label, test in tests:
            try:
                result = test()
            except Exception as e:
                result = f"{type(e).__name__}: {e}"
            if result is not True:
                failures.append((label, result))
        if failures:
            print(f"Question {q}: {len(tests) - len(failures)}/{len(tests)} tests passed")
            for label, reason in failures[:MAX_SHOWN]:
                if len(label) > 80: label = label[:77] + "..."
                if len(reason) > 150: reason = reason[:147] + "..."
                print(f"    FAILED {label}\n           {reason}")
            if len(failures) > MAX_SHOWN:
                print(f"    ... and {len(failures) - MAX_SHOWN} more failures")
        else:
            passed.append(q)
    print(f"\nTests passed for questions: {passed}")

run_lab5_tests()
```
